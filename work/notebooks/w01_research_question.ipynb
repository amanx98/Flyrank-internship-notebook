## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
url = "https://raw.githubusercontent.com/amanx98/Flyrank-internship-notebook/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)
```

```python
# Number 1: how many pages are even in scope (visible + old enough to judge)
in_scope = df[(df["impressions_90d"] > 0) & (df["content_age_days"] >= 90)]
print("Pages in scope:", len(in_scope), "of", len(df))

# Number 2: how many look declining right now (a proxy for "needs attention")
declining = in_scope[in_scope["trend_direction"] == "down"]
print("Declining:", len(declining), f"({len(declining)/len(in_scope):.1%} of in-scope pages)")

# Number 3: how many are thin but still pulling real demand (an "expand" candidate)
thin_but_visible = in_scope[(in_scope["word_count"] < 1200) & (in_scope["impressions_90d"] >= 250)]
print("Thin but visible:", len(thin_but_visible))
```

### Quick look at the data

**Sample:** 30,000 pages from the starter dataset. All 30,000 passed the in-scope filter (impressions_90d > 0, content_age_days ≥ 90) — this likely means the file was pre-built to only include this population, not that every page in the wider system qualifies. I haven't yet seen what was excluded (newer or zero-impression pages), so I can't say this sample is representative of the full inventory.

| Metric | Count | Share of in-scope |
|---|---:|---:|
| In-scope pages | 30,000 | 100% |
| Flagged declining (`trend_direction == "down"`) | 16,262 | 54.2% |
| Thin but visible (<1200 words, ≥250 impressions) | 82 | 0.3% |

**Why this matters, and what it doesn't prove yet:** refresh candidates outnumber expand candidates by roughly 200:1, which is why I'm scoping to those two actions rather than all five. But 54.2% flagged as "declining" is high enough to be suspicious rather than reassuring — it likely means the label fires on small, short-window wobbles as well as real sustained decline. Before I trust this number for anything beyond scoping, I need to check persistence and magnitude, not just take the flag at face value. Similarly, 82 examples is too thin a base to model statistically — useful as a rule-based flag, but not enough to train anything on without risking pure memorization.

## 4. The unit of analysis, as a real dataframe

One row represents one pseudonymized content item/page in the current trailing-90-day snapshot.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/amanx98/Flyrank-internship-notebook/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

in_scope = df[
    (df["impressions_90d"] > 0)
    & (df["content_age_days"] >= 90)
].copy()

in_scope["declining_proxy"] = (
    in_scope["trend_direction"] == "down"
).astype(int)

in_scope["expand_proxy"] = (
    in_scope["word_count"].lt(1200)
    & in_scope["impressions_90d"].ge(250)
).astype(int)

page_slice = in_scope[
    [
        "content_id",
        "client_id",
        "content_type",
        "impressions_90d",
        "clicks_90d",
        "sessions_90d",
        "word_count",
        "content_age_days",
        "days_since_last_update",
        "declining_proxy",
        "expand_proxy",
    ]
]

print("One row =", "one pseudonymized content item/page")
print("Shape:", page_slice.shape)

display(page_slice.head())
```

## 5. Why ML beats a fixed rule here
A fixed rule is useful as a baseline, but it may not be sufficient because review priority depends on several signals at once: impressions, clicks, sessions, word count, content age, update recency, content type, and missingness. One threshold cannot easily combine these signals or adapt to differences between clients and content types. A ranking model could learn a more useful ordering from observed past outcomes. However, I will keep ML only if it beats a transparent fixed-rule baseline on held-out data. Otherwise, the simpler rule is preferable.
